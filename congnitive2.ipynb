```python
import pandas as pd

# Create dataset
data = {
    'Employee_ID': [101, 102, 103, 104, 105],
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Edward'],
    'Department': ['HR', 'IT', 'IT', 'Marketing', 'Sales'],
    'Age': [29, 34, 41, 28, 38],
    'Salary': [50000, 70000, 65000, 55000, 60000],
    'Years_of_Experience': [4, 8, 10, 3, 12],
    'Joining_Date': ['2020-03-15', '2017-07-19', '2013-06-01', '2021-02-10', '2010-11-25'],
    'Gender': ['Female', 'Male', 'Male', 'Female', 'Male'],
    'Bonus': [5000, 7000, 6000, 4500, 5000],
    'Rating': [4.5, 4.0, 3.8, 4.7, 3.5]
}

df = pd.DataFrame(data)

# Save original dataset
df.to_csv('employees.csv', index=False)

# Read dataset
df = pd.read_csv('employees.csv')

# a) Shape
print("Shape:", df.shape)

# b) Summary
df.info()

# c) Descriptive statistics
print(df.describe())

# d) First 5 and last 3 rows
print("First 5 rows:")
print(df.head(5))

print("Last 3 rows:")
print(df.tail(3))

# e) Statistics
print("Average Salary:", df['Salary'].mean())
print("Total Bonus:", df['Bonus'].sum())
print("Youngest Age:", df['Age'].min())
print("Highest Rating:", df['Rating'].max())

# f) Sort by Salary descending
df = df.sort_values(by='Salary', ascending=False)

# g) Performance category
def performance_category(rating):
    if rating >= 4.5:
        return 'Excellent'
    elif rating >= 4.0:
        return 'Good'
    else:
        return 'Average'

df['Performance'] = df['Rating'].apply(performance_category)

# h) Missing values
print("Missing values:")
print(df.isnull().sum())

# i) Rename Employee_ID
df.rename(columns={'Employee_ID': 'ID'}, inplace=True)

# j) Employees with >5 years experience
print("Employees with more than 5 years experience:")
print(df[df['Years_of_Experience'] > 5])

# IT employees
print("IT Employees:")
print(df[df['Department'] == 'IT'])

# k) Add Tax
df['Tax'] = df['Salary'] * 0.10

# l) Save modified dataset
df.to_csv('modified_employees.csv', index=False)

print("Modified dataset saved successfully!")
```
