```python
import numpy as np

#(i)
arr = np.array([1, 2, 3, 6, 4, 5])

print("Reversed:", arr[::-1])


# (a)(ii)
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

values, counts = np.unique(x, return_counts=True)

max_count = np.max(counts)

most_frequent = values[counts == max_count]

print("\nFor x:")
print("Most frequent value:", most_frequent)

for value in most_frequent:
    print("Indices of", value, ":", np.where(x == value)[0])


# Q(b)(ii) Most frequent value and its indices
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

values, counts = np.unique(y, return_counts=True)

max_count = np.max(counts)

most_frequent = values[counts == max_count]

print("\nFor y:")
print("Most frequent values:", most_frequent)

for value in most_frequent:
    print("Indices of", value, ":", np.where(y == value)[0])
```

```python
import numpy as np

# Create 1-D array with 25 evenly spaced numbers from 10 to 100
mayank = np.linspace(10, 100, 25)

print("Array:")
print(mayank)

# Dimensions
print("Dimensions:", mayank.ndim)

# Shape
print("Shape:", mayank.shape)

# Total number of elements
print("Total elements:", mayank.size)

# Data type
print("Data type:", mayank.dtype)

#  bytes 
print("Total bytes:", mayank.nbytes)

# Transpose using reshape()
mayank_transpose = mayank.reshape(25, 1)

print("Transpose using reshape():")
print(mayank_transpose)

# Transpose using T attribute
print("Transpose using T:")
print(mayank.T)
```
