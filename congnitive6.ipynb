```python
## q1.
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

# a. Add 2°C
corrected_temperature = temperature + 2
print("Corrected temperature:", corrected_temperature)

# b. Celsius to Fahrenheit
fahrenheit = ((9/5) * corrected_temperature )+ 32
print("Fahrenheit:", fahrenheit)

# c. Readings greater than 32°C
greater_than_32 = corrected_temperature[corrected_temperature > 32]
print("Readings greater than 32°C:", greater_than_32)

# d. Count readings greater than 32°C
count = np.sum(corrected_temperature > 32)
print("Number of readings exceeding 32°C:", count)
```

```python
##q2.
import numpy as np

steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

# a. Total steps
print("Total steps:", np.sum(steps))

# b. Mean
print("Mean steps:", np.mean(steps))

# c. Maximum and minimum
print("Maximum:", np.max(steps))
print("Minimum:", np.min(steps))

# d. Total for each day
print("Total steps for each day:", np.sum(steps, axis=0))

# e. Total for each user
print("Total steps for each user:", np.sum(steps, axis=1))

# f. Position of maximum value
position = np.unravel_index(np.argmax(steps), steps.shape)
print("Position of maximum:", position)
print("Maximum value:", steps[position])
```

```python
## q4.
import numpy as np

# a. Create the original array
original = np.array([1, 2, 3, 4, 5, 6])
print("Original:", original)


# b. Create a slice from index 1 to 4
subset = original[1:5]
print("Subset:", subset)


# c. Modify the first element of subset
subset[0] = 999

print("After modifying subset:")
print("Original:", original)
print("Subset:", subset)


# d. Create another slice using copy()
copied = original[1:5].copy()

copied[0] = 500

print("\nAfter modifying copied array:")
print("Original:", original)
print("Copied:", copied)


# e. Create numbers from 1 to 12 and reshape into 3 x 4
matrix = np.arange(1, 13).reshape(3, 4)

print("\n3 x 4 Matrix:")
print(matrix)


# f. Indexing and slicing

# First row
print("\nFirst row:")
print(matrix[0, :])

# Last row
print("\nLast row:")
print(matrix[-1, :])

# Second column
print("\nSecond column:")
print(matrix[:, 1])

# Rows 1-2 and columns 2-3
print("\nRows 1-2 and columns 2-3:")
print(matrix[1:3, 2:4])


# g. Flatten and ravel
flat_array = matrix.flatten()
ravel_array = matrix.ravel()

print("\nFlatten:")
print(flat_array)

print("\nRavel:")
print(ravel_array)


# h. Modify ravel array
ravel_array[0] = 1000

print("\nAfter modifying ravel:")
print("Ravel:", ravel_array)
print("Original matrix:")
print(matrix)


# i. Modify flatten array
flat_array[1] = 2000

print("\nAfter modifying flatten:")
print("Flatten:", flat_array)
print("Original matrix:")
print(matrix)


# j. Display shape, ndim, size, dtype
print("\nMatrix properties:")
print("Shape:", matrix.shape)
print("Number of dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data type:", matrix.dtype)
```

```python
import numpy as np

# Given data
y = np.array([40, 65, 30, 85])

X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

# a. Shape and dimensions
print("Shape:", X.shape)
print("Dimensions:", X.ndim)

# b. Transpose
print("\nX.T:")
print(X.T)

# c. X.T @ X
XT_X = X.T @ X
print("\nX.T @ X:")
print(XT_X)

# d. Inverse
inverse = np.linalg.inv(XT_X)
print("\nInverse:")
print(inverse)

# e. OLS equation
XT_y = X.T @ y
beta = inverse @ XT_y

print("\nCoefficients:")
print(beta)

# f. Coefficients
print("\nSleep coefficient:", beta[0])
print("Activity coefficient:", beta[1])
print("Stress coefficient:", beta[2])

# g. Prediction
new_user = np.array([5, 40, 7])

prediction = new_user @ beta

print("\nPredicted assistance score:", prediction)
```
