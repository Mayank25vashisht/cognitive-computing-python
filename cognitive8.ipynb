```python
Topic = (
    "I recently started learning guitar because it matches my interests. "
    "I enjoy playing it whenever I have free time. "
    "Whenever I feel frustrated, I pick up my guitar and start playing, "
    "which helps me relax. "
    "Playing the guitar makes me feel calm, and I've never gotten bored of it. "
    "I learned more than ten songs in just three months. "
    "This achievement required a lot of effort, patience, and consistency."
)

import re
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.probability import FreqDist

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

text = re.sub(r'[^\w\s]', '', Topic.lower())
print("\n")
print("Lowercase and Punctuation removed")
print(text)



words = word_tokenize(Topic)
sentences = sent_tokenize(Topic)
print("\n")
print("Word Tokens:")
print(words)

print("\n")
print("Sentence Tokens:")
print(sentences)
print("\n")

print("Using split():")
print(Topic.split())
print("\n")

print("Using word_tokenize():")
print(word_tokenize(Topic))

print("\n")

#4
stop_words = set(stopwords.words("english"))

tokens = word_tokenize(text)
New_tokens = [
    word for word in tokens
    if word.lower() not in stop_words
]
print("After Stopwords Removal:")
print(New_tokens)

print("\n")
freq_dist = FreqDist(New_tokens)
print(freq_dist)

print("\n")

#v

print("\nWord Frequency Distribution:")
for word, frequency in freq_dist.items():
    print(word, ":", frequency)
```

```python
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

nltk.download('stopwords')
nltk.download('wordnet')


Topic = """I recently started learning guitar.
Not only does it align with my interests, but I also enjoy playing it.
Whenever I feel frustrated, I pick up my guitar and start playing it.
The feeling is relaxing, and I never get bored of it.
I learned more than 10 songs in just 3 months, which requires effort and consistency."""

#i
words = re.findall(r'\b[a-zA-Z]+\b', Topic.lower())
print("Words containing only alphabets:")
print(words)

#ii
stop_words = set(stopwords.words("english"))
filtered_words = [word for word in words if word not in stop_words]

print("\nAfter Stopword Removal:")
print(filtered_words)

#iii
ps = PorterStemmer()
stemmed_words = [ps.stem(word) for word in filtered_words]

print("\nStemmed Words:")
print(stemmed_words)

#iv
lemmatizer = WordNetLemmatizer()
lemmatized_words = [
    lemmatizer.lemmatize(word) for word in filtered_words
]

print("\nLemmatized Words:")
print(lemmatized_words)

#v
print("\nComparison:")
for word, stem, lemma in zip(filtered_words, stemmed_words, lemmatized_words):
    print(word, "-> Stem:", stem, "| Lemma:", lemma)

print("\nExplanation:")
print("Stemming reduces words by removing suffixes.")
print("Lemmatization uses a lexical database to find the base form of words.")
```

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
import pandas as pd

texts = [
    "I recently started learning guitar. I enjoy playing guitar because it aligns with my interests.",
    "Whenever I feel frustrated, I pick up my guitar and start playing. Playing guitar makes me feel relaxed and happy.",
    "I learned more than 10 songs in just 3 months. Learning guitar requires effort, patience and consistency."
]

#i
cv = CountVectorizer()
bow = cv.fit_transform(texts)

bow_df = pd.DataFrame(
    bow.toarray(),
    columns=cv.get_feature_names_out()
)

print("Bag of Words Representation:")
print(bow_df)

print("\n")
#ii
tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform(texts)

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf.get_feature_names_out()
)

print("TF-IDF Scores:")
print(tfidf_df.round(3))


print("\n")
#iii
features = tfidf.get_feature_names_out()

print("Top 3 Keywords from Each Text:")
print("\n")

for i in range(len(texts)):
    scores = tfidf_matrix[i].toarray()[0]
    top_indices = scores.argsort()[-3:][::-1]

    print(f"Text {i+1}:")
    for index in top_indices:
        print(features[index], ":", round(scores[index], 3))
```
