```python

import random as r
r.seed(1024170024)
num=[]
for i in range(100):
    num.append(r.randint(100,900))
print(num)

odd_num=[]
for i in num:
    if i%2!=0:
        odd_num.append(i)

print(odd_num, end=" ")
print(len(odd_num))

print("\n")


```

```python
even_num=[]
for i in num:
    if i%2==0:
        even_num.append(i)
print(even_num)
print(len(even_num))
print("\n")
```

```python
def is_prime(num):
    if num< 2:
        return False

    for i in range(2, int(num**0.5)+ 1):
        if num % i == 0:
            return False

    return True

prime=[i for i in num if is_prime(i)]
print(prime)
print(len(prime))
print("\n")


frequent = num[0]
highest_count=0

for i in num:
    count = num.count(i)
    if count > highest_count:
        highest_count = count
        frequent = i
print(frequent)
print("It occurs ",highest_count)
print("\n")
```

```python
rollno =[1,0,2,4,1,7,0,0,2,4]
A={ i*7 for i in rollno}
B={i*9 for i in rollno}
print(A)
print(B)
print('\n')
#i.
print(A.union(B))
print('\n')

#ii.
print(A.intersection(B))
print('\n')

#iii.
print(A.difference(B))
print('\n')


#iv
print("A-B", A.difference(B))
print("B-A", B.difference(A))
print('\n')

#v
print(A.symmetric_difference(B))
print('\n')

#vi
print("If A is subset of B")
print(A.issubset(B))
print("If B is superset of A")
print(B.issuperset(A))

print('\n')

#vii.
X=int(input("Enter a number: "))
A.discard(X)
print(A)
print("Cause discard will not raise error.")
```
